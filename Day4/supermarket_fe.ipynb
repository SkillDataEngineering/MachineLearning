# Supermarket Feature Engineering exercise
[Dataset on Kaggle](https://www.kaggle.com/datasets/aungpyaeap/supermarket-sales)

```python
import pandas as pd
```

```python
supermarket_df = pd.read_csv('supermarket_sales - Sheet1.csv')
```

## Time for the EDA - Exploratory Data Analysis

```python
supermarket_df.sample(10)
```

```python
supermarket_df.info()
```

```python
supermarket_df.corr(numeric_only=True)
```

```python
supermarket_df['Citybranch'] = supermarket_df['City']+'-'+supermarket_df['Branch']
supermarket_df.sample(20)
```

```python
customertypes = {'Normal':0, 'Member': 1}
supermarket_df['customer_code'] = supermarket_df['Customer type'].map(customertypes)
supermarket_df.sample(5)
```

```python
sex_codes = {'Male':0, 'Female': 1}
supermarket_df['gender_code'] = supermarket_df['Gender'].map(sex_codes)
supermarket_df.sample(5)
```

## Encoding categoricals with many categories
One-Hot Encoding

```python
from sklearn import preprocessing
enc = preprocessing.OneHotEncoder()
enc.fit(supermarket_df[['Product line']])
print(enc.categories_)
one_hot = enc.transform(supermarket_df[['Product line']]).toarray()
one_hot
```

```python
supermarket_df[['Electronic accessories', 'Fashion accessories',
       'Food and beverages', 'Health and beauty', 'Home and lifestyle',
       'Sports and travel' ]] = one_hot

supermarket_df.sample(6)
```

```python

target = supermarket_df['Rating']
feature_list = ['Unit price','Total', 'customer_code', 'gender_code','Electronic accessories', 'Fashion accessories','Food and beverages', 'Health and beauty', 'Home and lifestyle','Sports and travel']
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(supermarket_df[feature_list], target, test_size=0.2, random_state=42)

model = LinearRegression()

model.fit(X_train ,y_train)


predicted = model.predict(X_test)

```

```python
from sklearn import metrics
import numpy as np

mse = metrics.mean_squared_error(y_test, predicted)
print(f'MSE: {mse}')

print(f'RMSE: {np.sqrt(mse)}')

r2 = model.score(X_test, y_test)

print(f'R2 score: {r2}')
```
